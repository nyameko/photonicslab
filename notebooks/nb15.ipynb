# Lesson 15. Demonstration of 3D fully vectorial conical refraction in biaxial crystals (circularly polarized light on KGd)

Here we demonstrate how a modulated profile can be used as a source in subsequent profiles. This can be repeated indefinitely for simulation of an arbitrary number of self-thermal lensing profiles.

Specifically in this example we look at an incident Gaussian beam modulated by the pair of unequal adjascent Gaussians previously discused. Then we observe the self-phase modulation of this distorted beam on itself when impingent on a subsequent optic. Finally we compare a new Gaussian seed propagating through the modulation up to 40 cm, as well as the equivalent self-modulation upto 15 cm.

```python
import os, sys
module_path = os.path.abspath(os.path.join('../'))
if module_path not in sys.path:
    sys.path.append(module_path)

import importlib
import photonicslab
```

```python
mygaussQC = photonicslab.heateqn.HeatEqn2D(nt=200, nx=200, ny=200,hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
mygaussQC.add_source_from_hdf5(hdf5grp='nb1', hdf5table='seed_mod_convol')
mygaussQC.get_explicit_fd_params()
myfdsolverC = photonicslab.fdsolver.FDSolver2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
myfdsolverC.explicit_method()
myphasemodC = photonicslab.phasemod.ThermalLens(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
myphasemodC.parabolic_fit()
myphasemodC.phase()
myphasemodC.parabolic_fit(hdf5table='phase')
myviewC = photonicslab.visualize.Visualize2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
myviewC.steadystate_profiles(savepath='images/nb1c_steady_state.png')
#myviewC.transient_profiles(savepath='videos/nb1c_transient.mp4')

myphaseviewC = photonicslab.visualize.VisualizePhase2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
myphaseviewC.steadystate_profiles(savepath='images/nb1c_steady_state_phase.png', threshold=.9)
myphaseviewC.transient_profiles(savepath='videos/nb1c_transient_phase.mp4')

importlib.reload(photonicslab.visualize)
mybeamviewC = photonicslab.visualize.VisualizeSelfLensBeam2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1c')
mybeamviewC.steadystate_profiles(z_dist=10, savepath='images/nb1c_steady_state_beam_z10.png')
mybeamviewC.steadystate_profiles(z_dist=30, savepath='images/nb1c_steady_state_beam_z30.png')
mybeamviewC.steadystate_profiles(z_dist=50, savepath='images/nb1c_steady_state_beam_z50.png')
mybeamviewC.steadystate_profiles(z_dist=100, savepath='images/nb1c_steady_state_beam_z100.png')
mybeamviewC.transient_profiles(15, savepath='videos/nb1c_transient_beam_z15.mp4')
```

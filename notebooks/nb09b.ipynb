## 3D 2mm flattop-source phase modulated (VarOptic) Gaussian focusing

* We generated a transmission function corresponding to thermal-profile related phase modulation.


* We create a seed beam propagation object called `mybeam` and give the initial parameters:

    - Initial mesh grid spacing \ layout.
    - The time evolution matrix.
    - The transient solution to the heat equation.
    - The normalised (to fd coefficients) initial heat load.

* Generate and Plot seed beam profile.

* Generate and Play animation (loop) for temporal seed beam evolution at fixed z.

<div class="alert alert-block alert-info"> **Info:** Due to the numerics of the convolution function, it will show signs of breakdown near the boundaries. To circumvent this use a bigger meshgrid.</div>

```python
import os, sys
module_path = os.path.abspath(os.path.join('../'))
if module_path not in sys.path:
    sys.path.append(module_path)

import importlib
import photonicslab
```

```python
importlib.reload(photonicslab.heateqn)
mygauss3dQ = photonicslab.heateqn.HeatEqn3D(nx=81, ny=81, nz=51, nt=300, Lx=.7, Ly=.7, Lz=.5,\
                                            hdf5grp='nb11b', Qmax=400., gain_medium='tmyvo4', cavity_mode_waist=0.4)

mygauss3dQ.add_normalized_pump_gaussian(wp=0.1, alpha=1.25, flatness=3, q=400)
# Check the above units
mygauss3dQ.plot_source_profile(savepath='images/nb11b_sourceq.png', z_ind=-1)
mygauss3dQ.get_cn_fd_params()

importlib.reload(photonicslab.fdsolver)
myfdsolver3d = photonicslab.fdsolver.FDSolver3D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb11b')
myfdsolver3d.adi_method()

importlib.reload(photonicslab.visualize)
myview3d = photonicslab.visualize.Visualize3D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb11b')
myview3d.steadystate_profiles(savepath='images/nb11b_steady_state.png')
myview3d.transient_profiles(savepath='videos/nb11b_transient.mp4')

importlib.reload(photonicslab.phasemod)
myphasemod3d = photonicslab.phasemod.ThermalLens3D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb11b')
myphasemod3d.phase(hdf5table='temperature3d', hdf5dset_phase='phase3d',  hdf5dset_phase_collapsed='phase2d')
myphasemod3d.phase_heat(hdf5table='temperature3d', hdf5dset_phase='phase3d_heat',\
                        hdf5dset_phase_collapsed='phase2d_heat')

importlib.reload(photonicslab.visualize)
myphaseview3d = photonicslab.visualize.VisualizePhase3D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb11b',\
                                                       hdf5table='phase3d')
myphaseview3d.steadystate_profiles(savepath='images/nb11b_steady_state_phase.png')
myphaseview3d.transient_profiles(savepath='videos/nb11b_transient_phase.mp4')
```

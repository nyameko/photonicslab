## Solve the finite difference problem

```python
myfdsolverD = photonicslab.fdsolver.FDSolver2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
myfdsolverD.explicit_method()
myphasemodD = photonicslab.phasemod.ThermalLens(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
myphasemodD.parabolic_fit()
myphasemodD.phase()
myphasemodD.parabolic_fit(hdf5table='phase')
myviewD = photonicslab.visualize.Visualize2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
myviewD.steadystate_profiles(savepath='images/nb1d_steady_state.png')
myviewD.transient_profiles(savepath='videos/nb1d_transient.mp4')

myphaseviewD = photonicslab.visualize.VisualizePhase2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
myphaseviewD.steadystate_profiles(savepath='images/nb1d_steady_state_phase.png', threshold=.9)
myphaseviewD.transient_profiles(savepath='videos/nb1d_transient_phase.mp4')

mybeamviewD = photonicslab.visualize.VisualizeBeam2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
mybeamviewD.steadystate_profiles(z_dist=10, savepath='images/nb1d_steady_state_beam_z10.png')
mybeamviewD.steadystate_profiles(z_dist=30, savepath='images/nb1d_steady_state_beam_z30.png')
mybeamviewD.steadystate_profiles(z_dist=50, savepath='images/nb1d_steady_state_beam_z50.png')
mybeamviewD.steadystate_profiles(z_dist=100, savepath='images/nb1d_steady_state_beam_z100.png')
mybeamviewD.transient_profiles(10, savepath='videos/nb1d_transient_beam_z10.mp4')


mylgviewD = photonicslab.visualize.VisualizeBeam2D(l=1, phi=0.3,hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1d')
mylgviewD.transient_profiles(25, savepath='videos/nb1d_transient_beam_lg1_z10.mp4')
```

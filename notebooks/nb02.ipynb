# Demonstration of 2D source addition from an image file

Here we demonstrate how to add a source direcly from a file. The source can be repositioned or even added with temporal modulation. In this example we add a vortex vector beam.

```python
import os, sys
module_path = os.path.abspath(os.path.join('../'))
if module_path not in sys.path:
    sys.path.append(module_path)

import importlib
import photonicslab
```

```python
mygaussQimg = photonicslab.heateqn.HeatEqn2D(nt=100, nx=200, ny=200, hdf5filename='hdf5db/storage.hdf5',
                                         hdf5grp='nb1img')
mygaussQimg.add_source_from_image('reports/doughnut.png', center_x=-0.1)
mygaussQimg.add_source_from_image('reports/doughnut.png', center_x=0.1)
mygaussQimg.plot_source_profile(savepath='images/nb1img_sourceq.png')
mygaussQimg.get_explicit_fd_params()
from IPython.display import HTML, display
display(HTML("<table><tr><td><img src='images/doughnut.png' style='width: 250px'></td>\
             <td><img src='images/nb1img_sourceq.png' style='width: 400px'></td></tr></table>"))
```

```python
myfdsolver_img = photonicslab.fdsolver.FDSolver2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1img')
myfdsolver_img.explicit_method()
myphasemod_img = photonicslab.phasemod.ThermalLens(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1img')
myphasemod_img.parabolic_fit()
myphasemod_img.phase()
myphasemod_img.parabolic_fit(hdf5table='phase')
myview_img = photonicslab.visualize.Visualize2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1img')
myview_img.transient_profiles(savepath='videos/nb1img_transient.mp4')

myphaseview_img = photonicslab.visualize.VisualizePhase2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1img')
myphaseview_img.transient_profiles(savepath='videos/nb1img_transient_phase.mp4')

mybeamview_img = photonicslab.visualize.VisualizeBeam2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1img')
mybeamview_img.transient_profiles(30, savepath='videos/nb1img_transient_beam_z30.mp4')

mylgview_img = photonicslab.visualize.VisualizeBeam2D(l=1, phi=0.3,hdf5filename='hdf5db/storage.hdf5',
                                                  hdf5grp='nb1img')
mylgview_img.transient_profiles(30, savepath='videos/nb1img_transient_beam_lg1_z30.mp4',
                                hdf5_seed_intensity='seed_intensity_lg',
                                hdf5_seed_propagation='seed_propagation_lg')
```

## Dynamic and time-dependent heat sources 

Here we demonstration the additional of a moving, temporal source the the heat profile.

```python
import os, sys
module_path = os.path.abspath(os.path.join('../'))
if module_path not in sys.path:
    sys.path.append(module_path)

import importlib
import photonicslab
```

```python
mygaussQB = photonicslab.heateqn.HeatEqn2D(nt=200, nx=200, ny=200, hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
mygaussQB.add_gaussian(center_x=-0.1, nt_start=0, nt_end=50)
mygaussQB.add_gaussian(center_x=0.1, nt_start=51, nt_end=100)
mygaussQB.add_gaussian(center_y=-0.1, nt_start=101, nt_end=150)
mygaussQB.add_gaussian(center_y=0.1,nt_start=151, nt_end=200)
mygaussQB.get_explicit_fd_params()
myfdsolverB = photonicslab.fdsolver.FDSolver2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
myfdsolverB.explicit_method()
myphasemodB = photonicslab.phasemod.ThermalLens(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
myphasemodB.parabolic_fit()
myphasemodB.phase()
myphasemodB.parabolic_fit(hdf5table='phase')
myviewB = photonicslab.visualize.Visualize2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
myviewB.transient_profiles(savepath='videos/nb1b_transient.mp4')

myphaseviewB = photonicslab.visualize.VisualizePhase2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
myphaseviewB.transient_profiles(savepath='videos/nb1b_transient_phase.mp4')

mybeamviewB = photonicslab.visualize.VisualizeBeam2D(hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
mybeamviewB.transient_profiles(40, savepath='videos/nb1b_transient_beam_z40.mp4', hdf5table='seed_mod_convol_B')

mylgviewB = photonicslab.visualize.VisualizeBeam2D(l=1, phi=0.3,hdf5filename='hdf5db/storage.hdf5', hdf5grp='nb1b')
mylgviewB.transient_profiles(40, savepath='videos/nb1b_transient_beam_lg1_z40.mp4',
                             hdf5_seed_intensity='seed_intensity_lg',
                             hdf5_seed_propagation='seed_propagation_lg'))
```

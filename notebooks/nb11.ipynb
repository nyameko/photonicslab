## Validation of model by computing the transformation of a TEM$_{00}$ Gaussian beam by a lenslike medium, in which there are no stress effects.

* The medium is considered to be isotropic and with a spatially varying refractive index.


* We create a seed beam propagation object called `mybeam` and give the initial parameters:

    - Initial mesh grid spacing \ layout.
    - The time evolution matrix.
    - The transient solution to the heat equation.
    - The normalised (to fd coefficients) initial heat load.

* Laser wavelength considered to be $1.03\mu m$.

* Thickness of the medium is considered to be 4.5cm.

* Index of refraction distribution considered to be parabolic within the medium:
$
\begin{equation}
n = n_{0}\left(1 - 2\frac{r^2}{b^2}\right),
\end{equation}
$
where $r$ is the radial coordinate, $b = 1.2$ and $n_0 = 1.82$.

<div class="alert alert-block alert-info"> <b>Info:</b> Due to the numerics of the convolution function, it will show signs of breakdown near the boundaries. To circumvent this use a bigger meshgrid.</div>

```python
import os, sys
module_path = os.path.abspath(os.path.join('../'))
if module_path not in sys.path:
    sys.path.append(module_path)

import importlib
import photonicslab
```

```python
importlib.reload(photonicslab.isotropy)
importlib.reload(photonicslab.optomech)

# Generate isotropic GRINlens medium, and setup paraxial beam for regular lens propagation
myGRINLens = photonicslab.isotropy.GRINLens(nx=251, ny=251, nz=751, Lx=1.5, Ly=1.5, Lz=4.5, hdf5grp='grin')
myGRINLens.index_distr(n0=1.82, b=1.2)
f, h = myGRINLens.get_GRIN_focal_len()

print("Focal length: {0:.3f}m, and distance from medium surface to closest principle plane: {1:.3f}m"\
      .format(f, h))

#delete cavity_mode_waist=0.5 and other material variables that will be in gaineqn
# Make GRINLEns first simplest ancestor

mybeam = photonicslab.optomech.ParaxialWaveEqn(wavelen=1.064e-4, waist=500e-4, p=0, l=0, z=15, phi=0, \
                                               hdf5grp='grin', hdf5subgrp='tem00')
# TODO: ParaxialWaveEqn needs a constructor that uses existing Ax and Ay field functions
mybeam.adi_method()

#mybeam.free_space_adi_method(nz, Lz, hdf5grp, hdf5subgrp)
```
